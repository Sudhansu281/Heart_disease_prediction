```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
import matplotlib
matplotlib.use("TkAgg") 
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import joblib
```

```python
heart_data = pd.read_csv('heart_disease_data.csv')
```

```python
print(heart_data.head())
print(heart_data.shape)
print(heart_data.isnull().sum())
print(heart_data['target'].value_counts())
```

```python
X = heart_data.drop(columns='target', axis=1)
Y = heart_data['target']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, stratify=Y, random_state=2
)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = keras.Sequential([
    layers.Input(shape=(X_train_scaled.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(16, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
early_stop = keras.callbacks.EarlyStopping(
    monitor='val_loss', patience=10, restore_best_weights=True
)
```

```python
history = model.fit(
    X_train_scaled, Y_train,
    validation_split=0.2,
    epochs=100,
    batch_size=16,
    callbacks=[early_stop],
    verbose=1
)
```

```python
train_probs = model.predict(X_train_scaled)
train_preds = (train_probs > 0.5).astype(int)
training_data_accuracy = accuracy_score(train_preds, Y_train)
print('Accuracy on Training data : ', training_data_accuracy)
 
test_probs = model.predict(X_test_scaled)
test_preds = (test_probs > 0.5).astype(int)
test_data_accuracy = accuracy_score(test_preds, Y_test)
print('Accuracy on Test data : ', test_data_accuracy)
```

```python
model.save("heart_model_ann.keras")
joblib.dump(scaler, "heart_scaler.pkl")
 
print("Saved: heart_model_ann.keras, heart_scaler.pkl")
```
